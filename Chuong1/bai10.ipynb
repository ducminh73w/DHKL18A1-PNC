```python
import math
class Diem:
    def __init__(self, x=0, y=0):
        self.x = x
        self.y = y
        print(f"-> Tạo Điểm tại ({self.x}, {self.y})")

    def __str__(self):
        return f"Tâm: ({self.x}, {self.y})"

class Elip(Diem):
    def __init__(self, x, y, ban_truc_lon, ban_truc_be):
        super().__init__(x, y)
        
        self.a = ban_truc_lon
        self.b = ban_truc_be
        
        if self.a <= 0 or self.b <= 0:
            raise ValueError("Bán trục phải là số dương.")
        if self.a < self.b:
            self.a, self.b = self.b, self.a 
            
        print(f"-> Tạo Elip: {self.a} (lớn), {self.b} (bé)")

    def tinh_dien_tich(self):
        """Tính diện tích của Elip: S = Pi * a * b"""
        return math.pi * self.a * self.b

    def __str__(self):
        thong_tin_tam = super().__str__()
        return f"Elip: {thong_tin_tam}, Bán trục lớn/bé: {self.a}/{self.b}, Diện tích: {self.tinh_dien_tich():.2f}"

class Tron(Elip):
    """Lớp mô tả hình Tròn, kế thừa từ Elip (a = b = bán kính)."""
    def __init__(self, x, y, ban_kinh):
        super().__init__(x, y, ban_kinh, ban_kinh)
        self.r = ban_kinh
        print(f"-> Tạo Hình Tròn: Bán kính {self.r}")
        
    def __str__(self):
        thong_tin_tam = Diem.__str__(self) 
        return f"Hình Tròn: {thong_tin_tam}, Bán kính: {self.r}, Diện tích: {self.tinh_dien_tich():.2f}"

```

```python
print("="*40)
print(" ktra : Điểm, Elip, Tròn")
print("="*40)
try:
    elip_1 = Elip(x=int(input(" nhập số nguyên: ")), y=int(input("nhập số nguyên: ")), ban_truc_lon=8, ban_truc_be=5)
    print(elip_1)
    
    elip_2 = Elip(0, 0, 4, 10) 
    print(elip_2)
    
except ValueError as e:
    print(f"Lỗi khởi tạo Elip: {e}")

try:
    print("\n--- Test Hình Tròn ---")
    tron_1 = Tron(x=int(input(" nhập số nguyên: ")), y=int(input(" nhập số nguyên: ")), ban_kinh=int(input(" nhập số nguyên: ")))
    print(tron_1)
    
    dien_tich_tron = tron_1.tinh_dien_tich()
    print(f"Kiểm tra tính diện tích: {dien_tich_tron:.2f} (Pi * 6^2 = {math.pi * 36:.2f})")

except ValueError as e:
    print(f"Lỗi khởi tạo Hình Tròn: {e}")
```
