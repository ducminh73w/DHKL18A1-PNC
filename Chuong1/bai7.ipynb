```python
class Date:
    def __init__(self, a=1, b=1, c=2000):
        self.day = a
        self.month = b
        self.year = c

    def display(self):
        print(f"{self.day:02d}/{self.month:02d}/{self.year}")

    def next(self):
        import datetime
        cur = datetime.date(self.year, self.month, self.day)
        next_day = cur + datetime.timedelta(days=1)
        return Date(next_day.day, next_day.month, next_day.year)
```

```python
def main():
    print(" kiểm tra lớp date")
    a = int(input("Nhập ngày: "))
    b = int(input("Nhập tháng: "))
    c = int(input("Nhập năm: "))
    today = Date(a, b, c)
    print("\nNgày hiện tại:", end=" ")
    today.display()
    tomorrow = today.next()
    print("Ngày tiếp theo:", end=" ")
    tomorrow.display()


if __name__ == "__main__":
    main()
```
