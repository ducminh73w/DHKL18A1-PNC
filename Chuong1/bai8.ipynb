```python
class Date:
    def __init__(self, d=1, m=1, y=2000):
        self.day = d
        self.month = m
        self.year = y

    def nhap(self):
        self.day = int(input("  Ngày: "))
        self.month = int(input("  Tháng: "))
        self.year = int(input("  Năm: "))

    def display(self):
        print(f"{self.day:02d}/{self.month:02d}/{self.year}", end="")

```

```python
class Employee:
    def __init__(self):
        self.name = ""
        self.birth = Date()
        self.join = Date()

    def nhap(self):
        self.name = input("Nhập họ tên nhân viên: ")
        print("Nhập ngày sinh:")
        self.birth.nhap()
        print("Nhập ngày vào công ty:")
        self.join.nhap()

    def in_thong_tin(self):
        print("\n--- Thông tin nhân viên ---")
        print(f"Họ tên: {self.name}")
        print("Ngày sinh: ", end=""); self.birth.display(); print()
        print("Ngày vào công ty: ", end=""); self.join.display(); print()
```

```python
ds_nv = []
n = int(input("Nhập số lượng nhân viên: "))
for i in range(n):
    print(f"\n--- Nhập nhân viên thứ {i+1} ---")
    nv = Employee()
    nv.nhap()
    ds_nv.append(nv)

print("\n DS NHÂN VIÊN")
for nv in ds_nv:
    nv.in_thong_tin()
```
