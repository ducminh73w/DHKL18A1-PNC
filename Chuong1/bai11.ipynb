```python
import math

# Lớp Tam giác cơ bản
class Triangle:
    def __init__(self, a, b, c):
        self.sides = [a, b, c]
        if not self.is_valid():
            raise ValueError("Ba cạnh không tạo thành tam giác hợp lệ!")

    def is_valid(self):
        a, b, c = self.sides
        return (a + b > c) and (a + c > b) and (b + c > a)

    def perimeter(self):
        return sum(self.sides)

    def area(self):
        a, b, c = self.sides
        s = self.perimeter() / 2
        return math.sqrt(s * (s - a) * (s - b) * (s - c))

    def display_sides(self):
        print("Các cạnh:", self.sides)


# Tam giác cân
class IsoscelesTriangle(Triangle):
    def __init__(self, a, b, c):
        super().__init__(a, b, c)
        if not self.is_isosceles():
            raise ValueError("Không phải tam giác cân!")

    def is_isosceles(self):
        a, b, c = self.sides
        return a == b or b == c or a == c


#Tam giác vuông
class RightTriangle(Triangle):
    def __init__(self, a, b, c):
        super().__init__(a, b, c)
        if not self.is_right():
            raise ValueError("Không phải tam giác vuông!")

    def is_right(self):
        a, b, c = sorted(self.sides)
        return math.isclose(a*a + b*b, c*c)


# Tam giác đều
class EquilateralTriangle(Triangle):
    def __init__(self, a):
        super().__init__(a, a, a)

    def area(self):
        a = self.sides[0]
        return (math.sqrt(3) / 4) * a * a
```

```python
def main():
    print("=== CHƯƠNG TRÌNH XÁC ĐỊNH LOẠI TAM GIÁC ===")

    try:

        a = float(input("Nhập cạnh a: "))
        b = float(input("Nhập cạnh b: "))
        c = float(input("Nhập cạnh c: "))
        if a == b == c:
            tg = EquilateralTriangle(a)
            print("→ Đây là tam giác đều.")
        elif a == b or b == c or a == c:
            try:
                tg = RightTriangle(a, b, c)
                print("→ Đây là tam giác vuông cân.")
            except ValueError:
                tg = IsoscelesTriangle(a, b, c)
                print("→ Đây là tam giác cân.")
        else:
            try:
                tg = RightTriangle(a, b, c)
                print("→ Đây là tam giác vuông.")
            except ValueError:
                tg = Triangle(a, b, c)
                print("→ Đây là tam giác thường.")

        print(f"Các cạnh: {tg.sides}")
        print(f"Chu vi: {tg.perimeter():.2f}")
        print(f"Diện tích: {tg.area():.2f}")

    except ValueError as e:
        print("Lỗi:", e)


if __name__ == "__main__":
    main()
```
