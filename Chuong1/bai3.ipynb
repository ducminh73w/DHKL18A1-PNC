```python
from math import gcd

class PhanSo:
    def __init__(self, tu=0, mau=1):
        self.tu = tu
        self.mau = mau

    def hop_le(self):
        return self.mau != 0

    def nhap(self):
        self.tu = int(input("Nhập tử số: "))
        self.mau = int(input("Nhập mẫu số: "))
        if not self.hop_le():
            print("Phân số không hợp lệ! Mẫu phải khác 0.")
            self.mau = 1

    def rut_gon(self):
        ucln = gcd(self.tu, self.mau)
        self.tu //= ucln
        self.mau //= ucln

    def in_phan_so(self):
        self.rut_gon()
        print(f"{self.tu}/{self.mau}")
```
