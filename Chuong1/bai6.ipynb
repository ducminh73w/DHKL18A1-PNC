```python
class Stack:
    def __init__(self, size):
        self.size = size
        self.data = []

    def isEmpty(self):
        return len(self.data) == 0

    def isFull(self):
        return len(self.data) == self.size

    def push(self, value):
        if self.isFull():
            print("Ngăn xếp đã đầy!")
        else:
            self.data.append(value)

    def pop(self):
        if self.isEmpty():
            print("Ngăn xếp rỗng!")
        else:
            return self.data.pop()
```

```python
class StackCount(Stack):
    def count(self):
        return len(self.data)
```

```python
class StackPrint(StackCount):
    def print(self):
        print("Nội dung ngăn xếp:", self.data)
```

```python
def main():
    print("=== CHƯƠNG TRÌNH KIỂM TRA NGĂN XẾP ===")

    # Nhập kích thước ngăn xếp
    n = int(input("Nhập kích thước ngăn xếp: "))
    stack = StackPrint(n)

    while True:
        print("\n--- MENU ---")
        print("1. Thêm phần tử (Push)")
        print("2. Lấy phần tử ra (Pop)")
        print("3. In nội dung ngăn xếp")
        print("4. Đếm số phần tử trong ngăn xếp")
        print("0. Thoát")

        choice = input("Chọn chức năng: ")

        if choice == "1":
            val = input("Nhập giá trị cần thêm: ")
            stack.push(val)

        elif choice == "2":
            x = stack.pop()
            if x is not None:
                print("Phần tử vừa lấy ra:", x)

        elif choice == "3":
            stack.print()

        elif choice == "4":
            print("Số phần tử hiện có:", stack.count())

        elif choice == "0":
            print("Kết thúc chương trình!")
            break

        else:
            print("Lựa chọn không hợp lệ!")


if __name__ == "__main__":
    main()
```
