```python
class Stack:
    def __init__(self, size):
        self.size = size
        self.data = []

    def isEmpty(self):
        return len(self.data) == 0

    def isFull(self):
        return len(self.data) == self.size

    def push(self, value):
        if self.isFull():
            print("Ngăn xếp đã đầy!")
        else:
            self.data.append(value)

    def pop(self):
        if self.isEmpty():
            print("Ngăn xếp rỗng!")
        else:
            return self.data.pop()
```

```python
class StackCount(Stack):
    def count(self):
        return len(self.data)
```
