```python
class DaGiac:
    def __init__(self, so_canh):
        self.so_canh = so_canh
        self.ds_canh = []

    def nhap(self):
        """Nhập độ dài các cạnh của đa giác"""
        self.ds_canh = []
        print(f"Nhập {self.so_canh} cạnh của đa giác:")
        for i in range(self.so_canh):
            canh = float(input(f"  Cạnh {i+1}: "))
            self.ds_canh.append(canh)

    def tinh_chu_vi(self):
        """Tính chu vi đa giác"""
        return sum(self.ds_canh)

    def in_thong_tin(self):
        print(f"Số cạnh: {self.so_canh}")
        print("Các cạnh:", self.ds_canh)
        print(f"Chu vi: {self.tinh_chu_vi()}")

```

```python
class HinhBinhHanh(DaGiac):
    def __init__(self):
        super().__init__(4)   # hình bình hành có 4 cạnh
        self.day = 0
        self.canh_ben = 0
        self.chieu_cao = 0

    def nhap(self):
        """Nhập cạnh đáy, cạnh bên và chiều cao"""
        print("=== NHẬP HÌNH BÌNH HÀNH ===")
        self.day = float(input("Nhập cạnh đáy: "))
        self.canh_ben = float(input("Nhập cạnh bên: "))
        self.chieu_cao = float(input("Nhập chiều cao: "))
        self.ds_canh = [self.day, self.canh_ben, self.day, self.canh_ben]

    def tinh_dien_tich(self):
        """Tính diện tích hình bình hành"""
        return self.day * self.chieu_cao

    def in_thong_tin(self):
        print("\n--- Thông tin hình bình hành ---")
        print(f"Cạnh đáy: {self.day}")
        print(f"Cạnh bên: {self.canh_ben}")
        print(f"Chiều cao: {self.chieu_cao}")
        print(f"Chu vi: {self.tinh_chu_vi()}")
        print(f"Diện tích: {self.tinh_dien_tich()}")
```

```python
class HinhChuNhat(HinhBinhHanh):
    def __init__(self):
        super().__init__()
        self.chieu_dai = 0
        self.chieu_rong = 0

    def nhap(self):
        """Nhập chiều dài và chiều rộng"""
        print("=== NHẬP HÌNH CHỮ NHẬT ===")
        self.chieu_dai = float(input("Nhập chiều dài: "))
        self.chieu_rong = float(input("Nhập chiều rộng: "))
        self.ds_canh = [self.chieu_dai, self.chieu_rong,
                        self.chieu_dai, self.chieu_rong]

    def tinh_dien_tich(self):
        """Tính diện tích hình chữ nhật"""
        return self.chieu_dai * self.chieu_rong

    def in_thong_tin(self):
        print("\n--- Thông tin hình chữ nhật ---")
        print(f"Chiều dài: {self.chieu_dai}")
        print(f"Chiều rộng: {self.chieu_rong}")
        print(f"Chu vi: {self.tinh_chu_vi()}")
        print(f"Diện tích: {self.tinh_dien_tich()}")
```

```python
hbh = HinhBinhHanh()
hbh.nhap()
hbh.in_thong_tin()

hcn = HinhChuNhat()
hcn.nhap()
hcn.in_thong_tin()
```
